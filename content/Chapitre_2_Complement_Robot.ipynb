# Chapitre 2 — Complément : exercices avec un robot

Sur France-IOI, plusieurs exercices du chapitre 2 (« Transport d'eau », « Vendanges »...) demandent de diriger un petit robot le long d'une rue, avec des instructions comme `gauche()`, `droite()`, `ramasser()` et `deposer()`. Sur le vrai site, ces instructions viennent d'un module tout prêt, avec `from robot import *`.

Ce notebook tourne sur un vrai noyau Python : il n'existe pas de vrai module `robot` ici, alors on définit nous-mêmes ces quatre instructions, avec exactement le même nom, dans la cellule ci-dessous. Une fois cette cellule exécutée, tu peux les utiliser exactement comme sur France-IOI.

**Exécute la cellule suivante en premier, avant de commencer les exercices :**

```python
position = 0
a_un_objet = False
nb_deplacements = 0

def demarrer(depart=0):
    """Remet le robot à une position de départ, sans objet."""
    global position, a_un_objet, nb_deplacements
    position = depart
    a_un_objet = False
    nb_deplacements = 0

def gauche():
    """Déplace le robot d'une case vers la gauche."""
    global position, nb_deplacements
    position -= 1
    nb_deplacements += 1

def droite():
    """Déplace le robot d'une case vers la droite."""
    global position, nb_deplacements
    position += 1
    nb_deplacements += 1

def ramasser():
    """Le robot ramasse l'objet à ses pieds."""
    global a_un_objet
    a_un_objet = True

def deposer():
    """Le robot dépose l'objet qu'il transportait."""
    global a_un_objet
    a_un_objet = False

def afficher_etat():
    """Affiche la position et l'état actuels du robot."""
    print(f"Position : {position}  |  Objet transporté : {'oui' if a_un_objet else 'non'}  |  Déplacements effectués : {nb_deplacements}")

def afficher_rue(taille=30):
    """Affiche une représentation de la rue avec le robot (R)."""
    rue = ["-"] * taille
    if 0 <= position < taille:
        rue[position] = "R"
    print("".join(rue))
```

💡 Remarque : contrairement à `print()`, ces instructions ne montrent rien par défaut — exactement comme un vrai robot, dont on ne voit le résultat qu'en le regardant. Utilise `afficher_etat()` ou `afficher_rue()` quand tu veux vérifier où il en est.
