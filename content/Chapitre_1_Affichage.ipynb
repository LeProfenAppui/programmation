## 1. Qu'est-ce qu'un programme ?

Un **programme** est une suite d'**instructions** que l'ordinateur exécute les unes après les autres, **dans l'ordre où elles sont écrites** — du début à la fin, comme une recette de cuisine qu'on suit étape par étape.

L'ordinateur ne devine jamais ce que tu veux faire : il exécute exactement ce qui est écrit, ni plus ni moins. Si tu changes l'ordre des instructions, le résultat peut changer.

### 🏴‍☠️ Un rappel de Pyrates

Dans Pyrates, quand tu ajoutais un bloc comme `avancer()`, tu voyais **immédiatement** le pirate se déplacer à l'écran : le jeu affichait lui-même, tout seul, le résultat de chaque instruction. Tu n'avais jamais besoin de demander à voir ce qui se passait — c'était déjà sous tes yeux, grâce au décor du jeu.

En Python, ce n'est plus du tout le cas. Un programme peut s'exécuter parfaitement, faire des calculs compliqués... et **n'afficher absolument rien**, si on ne le lui demande pas explicitement. Il n'y a pas de plateau de jeu qui montre le résultat à ta place.

C'est exactement là qu'intervient `print()` : c'est lui qui joue le rôle que jouait l'écran de Pyrates — **montrer ce qui se passe à l'intérieur du programme**.

## 6bis. `print()` avec une virgule ou avec `+` : quelle différence ?

Tu as maintenant vu les deux façons d'afficher plusieurs éléments ensemble. Elles se ressemblent, mais ne fonctionnent pas pareil.

**Avec une virgule**, Python affiche chaque élément l'un après l'autre, en ajoutant **automatiquement un espace** entre eux — et les éléments n'ont pas besoin d'être du même type (texte et nombre peuvent se mélanger sans rien faire de spécial).

````python
age = 15
print("Âge :", age)
````

**Avec `+`**, tu ne donnes qu'un seul élément à `print()` : le résultat d'un assemblage fait *avant* l'affichage. Aucun espace n'est ajouté automatiquement, et surtout, **tous les éléments doivent être du même type** — il faut donc convertir un nombre avec `str()` avant de l'assembler à du texte.

🎯 **`str()`** est une instruction qui **convertit un nombre en une chaîne de caractères** (c'est-à-dire en texte). Une fois converti, ce texte peut être assemblé avec `+` comme n'importe quel autre morceau de texte.

````python
age = 15
print("Âge : " + str(age))
````

⚠️ Remarque bien l'espace manquant si on l'oublie avec `+` :

````python
print("Salut" + "Théo")
print("Salut", "Théo")
````

| | Virgule `,` | Plus `+` |
|---|---|---|
| Espace automatique | Oui | Non |
| Mélange texte/nombre | Oui, directement | Non, il faut `str()` d'abord |
| Ce que reçoit `print()` | Plusieurs éléments séparés | Un seul élément déjà assemblé |

En résumé : la virgule est plus simple et plus tolérante ; `+` demande plus de rigueur, mais donne un contrôle total sur ce qui est collé exactement (utile par exemple pour éviter l'espace automatique).

## 7. Les erreurs de syntaxe courantes

⚠️ Un guillemet oublié ou mal fermé provoque une **erreur de syntaxe** (`SyntaxError`). C'est l'erreur la plus fréquente en tout début d'apprentissage — et bonne nouvelle : Python te dit toujours à quelle ligne elle se trouve.

Exécute la cellule suivante : elle contient volontairement une erreur.

````python
print("Ceci va provoquer une erreur)
````

Tu obtiens un message d'erreur (`SyntaxError`). C'est normal : il manque un guillemet fermant. Corrige-le directement dans la cellule ci-dessus et réexécute jusqu'à ce que ça fonctionne, avant de continuer.

⚠️ Une autre erreur fréquente : que faire si le texte à afficher contient **lui-même** un guillemet (ou une apostrophe), qui n'est ni ouvrant ni fermant ? Sans précaution, Python croit que le texte s'arrête à ce guillemet-là, et le reste de la ligne provoque une erreur. La solution est d'ajouter un **antislash `\`** juste avant, pour indiquer à Python de l'afficher tel quel plutôt que de le traiter comme la fin du texte.

````python
print("Il a dit : \"Salut\" à tout le monde")
````

## 🎁 Pour aller plus loin : les modules

🎯 **À quoi ça sert ?** Un **module** est une sorte de boîte à outils supplémentaire : il contient des instructions toutes prêtes, écrites par d'autres, que tu peux utiliser dans ton programme sans avoir à les réécrire toi-même.

📝 **Syntaxe :** avant d'utiliser un module, il faut d'abord l'**importer** avec le mot-clé `import`, tout en haut du programme. Ensuite, on accède à ses outils avec `nom_du_module.quelque_chose`.

Voici un exemple avec le module `random`, qui fournit des outils liés au hasard — ici, un nombre décimal tiré au hasard entre 0 et 1 :

````python
import random

print(random.random())
````

Réexécute plusieurs fois la cellule ci-dessus (Maj + Entrée) : le résultat change à chaque fois, c'est normal — c'est bien tout l'intérêt du hasard !

✔️ **À retenir :**
- `import nom_du_module` se place en général tout en haut du programme.
- On utilise ensuite ses outils avec `nom_du_module.quelque_chose`.
- Tu rencontreras d'autres modules plus tard dans l'année, pour d'autres besoins.
