## 1. Le problème de la répétition

Imagine que tu doives afficher le mot « Bonjour » cinq fois. Tu pourrais écrire :

````python
print("Bonjour")
print("Bonjour")
print("Bonjour")
print("Bonjour")
print("Bonjour")
````

Ça fonctionne... mais imagine devoir le faire 100 fois, ou 1000 fois ! Recopier la même instruction encore et encore n'est vraiment pas pratique. C'est exactement le problème que résout la **boucle**.

## 2. La boucle `for`

🎯 **À quoi ça sert ?** La boucle `for` répète un bloc d'instructions un nombre de fois **connu à l'avance**, sans avoir à le recopier.

📝 **Syntaxe :** on utilise `for`, une variable, `in range(N)`, puis **deux-points**. Le résultat exact du chapitre précédent, mais en une boucle :

````python
for i in range(5):
    print("Bonjour")
````

`range(5)` veut dire « 5 fois ». En réalité, `range(5)` correspond aux valeurs `0, 1, 2, 3, 4` — c'est-à-dire de 0 à N-1 — ce qui fait bien N valeurs (N termes) au total, même si on ne les affiche pas encore dans ce chapitre. Essaie de changer ce nombre pour répéter plus ou moins souvent.

💡 Tu remarques que la boucle utilise un nom de variable (`i`) juste après `for`, même si on ne s'en sert pas ici. Cette variable existe bel et bien à chaque tour — on apprendra à l'utiliser dans un chapitre plus tard, quand ce sera utile pour compter ou traiter plusieurs valeurs saisies. Pour l'instant, retiens simplement qu'elle doit être présente dans la syntaxe.

### 🏴‍☠️ Un rappel de Pyrates

Tu avais déjà utilisé une boucle `for` dans Pyrates, avec `range()` exactement comme ici — par exemple pour répéter `avancer()`, ou pour enchaîner plusieurs instructions à chaque tour (comme donner un coup d'épée puis avancer, au niveau 2). ## 3. Le corps de la boucle et l'indentation

Tout ce qui est **décalé vers la droite** (indenté) juste après le `for ... :` fait partie du bloc répété — on l'appelle le **corps de la boucle**. Il peut contenir plusieurs instructions, qui seront toutes rejouées à chaque tour.

````python
for i in range(3):
    print("Ligne 1")
    print("Ligne 2")
````

⚠️ **Attention :** Python est très strict sur l'indentation. Regarde ce qui se passe si la deuxième ligne n'est **pas** indentée : elle sort du corps de la boucle, et n'est donc plus répétée.

````python
for i in range(3):
    print("Ligne 1")
print("Ligne 2")
````

Compare bien les deux résultats ci-dessus : `Ligne 1` est répétée 3 fois dans les deux cas, mais `Ligne 2` ne l'est que dans le premier exemple.

## 4. `range()` avec un début et une fin

Jusqu'ici, `range(N)` commençait toujours à 0 et répétait N fois. On peut aussi préciser un **début** différent, en donnant **deux** nombres à `range()`.

````python
for i in range(2, 6):
    print("Bonjour")
````

Compte le nombre de fois où `Bonjour` s'affiche : 4 fois. `range(2, 6)` répète (6 − 2) = 4 fois — la différence entre les deux nombres donnés. Ces 4 tours correspondent aux valeurs `2, 3, 4, 5` (la fin, 6, n'est jamais atteinte).

## 5. `range()` avec un pas

On peut ajouter un **troisième** nombre à `range()` : le **pas**, c'est-à-dire de combien on avance à chaque tour. Par défaut, le pas vaut 1 — mais on peut le changer.

````python
for i in range(0, 10, 2):
    print("Beep")
````

Avec un pas de 2 entre 0 et 10, on obtient 5 répétitions (au lieu de 10 avec un pas de 1) : le pas divise le nombre de tours. Ces 5 tours correspondent aux valeurs `0, 2, 4, 6, 8`.

## 6. Les boucles imbriquées

On peut mettre une boucle `for` **à l'intérieur** d'une autre boucle `for`. On dit qu'elles sont **imbriquées**. La boucle intérieure recommence entièrement à chaque tour de la boucle extérieure.

````python
for i in range(3):
    for j in range(2):
        print("Ding")
````

Remarque bien l'indentation : le `for j` est indenté sous `for i`, et le `print()` est indenté encore une fois de plus, sous `for j`. Pour chacun des 3 tours extérieurs, les 2 tours intérieurs sont parcourus en entier : cela fait 3 × 2 = 6 « Ding » affichés au total.

On peut aussi placer une instruction directement dans la boucle **extérieure**, mais **en dehors** de la boucle intérieure : elle s'exécute alors une fois par tour extérieur seulement, pas une fois par tour intérieur.

````python
for i in range(3):
    for j in range(4):
        print("Beep")
    print("---")
````

Regarde bien l'indentation : `print("---")` est aligné avec `for j`, pas avec `print("Beep")`. Il fait donc partie du corps de la boucle **extérieure**, mais pas de la boucle intérieure. Résultat : `Beep` s'affiche 4 fois à chaque tour extérieur (12 fois au total), mais `---` ne s'affiche qu'une seule fois par tour extérieur (3 fois au total).

## 7. Les erreurs fréquentes avec les boucles

⚠️ Deux erreurs reviennent très souvent avec les boucles :
- Oublier les **deux-points** `:` à la fin de la ligne `for`.
- Une **indentation incohérente**, qui change silencieusement ce qui est répété (sans forcément provoquer d'erreur visible).

Exécute la cellule suivante : elle contient volontairement une erreur.

````python
for i in range(3)
    print("Test")
````

Tu obtiens une `SyntaxError` : il manque les deux-points après `range(3)`. Corrige-la directement ci-dessus et réexécute jusqu'à ce que ça fonctionne, avant de continuer.

- `range(N)` veut dire « N fois ».

## 🚀 Exercices supplémentaires (pour aller plus loin)

`print()` accepte un réglage optionnel `end=`, qui remplace le retour à la ligne automatique de fin (par défaut, `end` vaut toujours `"\n"`, un retour à la ligne). En mettant `end=""`, on empêche ce retour à la ligne : le prochain `print()` s'affichera juste à la suite, sur la même ligne.

````python
print("Bonjour", end="")
print("Bonjour", end="")
````

Les deux `Bonjour` sont collés, sans espace ni retour à la ligne entre eux. C'est très utile dans une boucle, pour construire une ligne petit à petit, caractère par caractère ou bloc par bloc, avant de finalement passer à la ligne avec un `print()` tout seul, sans argument.
